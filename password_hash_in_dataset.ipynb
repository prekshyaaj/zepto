```python
import pandas as pd
import bcrypt

# Load your users dataset
df = pd.read_csv("/Users/prekshyajoshi/Downloads/users.csv")

# Function to hash password
def hash_password(password):
    return bcrypt.hashpw(password.encode(), bcrypt.gensalt()).decode()

# Use SAME password for all users (for demo)
COMMON_PASSWORD = "test123"

# Replace TEMP with real hashes
df["password_hash"] = df["password_hash"].apply(
    lambda x: hash_password(COMMON_PASSWORD)
)

# Save updated file
df.to_csv("users_hash.csv", index=False)

print("✅ Passwords hashed successfully!")
```

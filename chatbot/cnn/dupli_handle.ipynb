```python
#checking whts the extensions of diffr images in our dataset

import os
from collections import Counter

DATASET_PATH = "/Users/prekshyajoshi/instacart_recommender_project/zepto_cnn_dataset_final/good"

format_counter = Counter()

for root, dirs, files in os.walk(DATASET_PATH):
    for file in files:
        ext = os.path.splitext(file)[1].lower()  # get extension like .jpg
        
        if ext:  # ignore files without extension
            format_counter[ext] += 1

print("\n📊 Image format distribution:\n")
for ext, count in format_counter.items():
    print(f"{ext}: {count}")
```

```python
#converting all diffr img extensions to jpg for uniformity and renaming files as a1,a2,.. 

import os
from PIL import Image  #using pillow lib since python cant work directly w images,itll help to conv to rgb,jpg also

DATASET_PATH = "/Users/prekshyajoshi/instacart_recommender_project/zepto_cnn_dataset_final/good"

temp_files = []
count = 1

# STEP 1: Convert everything to JPG with temp names
for root, dirs, files in os.walk(DATASET_PATH):
    for file in files:
        filepath = os.path.join(root, file)

        try:
            with Image.open(filepath) as img:  #opening file via pillow as img
                img = img.convert("RGB")       #since pretrained models like imgnet expects rgb format only

                temp_name = f"temp_{count}.jpg"
                temp_path = os.path.join(DATASET_PATH, temp_name)

                img.save(temp_path, "JPEG", quality=95)

                temp_files.append(temp_path)
                count += 1

        except Exception as e:
            print(f"Skipped {filepath}: {e}")

# STEP 2: Delete ALL original files since rn we have original .webp,.avif,.png along w their converted .jpg files so we remove the og ones
for root, dirs, files in os.walk(DATASET_PATH):
    for file in files:
        if not file.startswith("temp_"):  # keep only temp files i.e. the new ones
            os.remove(os.path.join(root, file))

# STEP 3: Rename temp files to final names
final_count = 1

for file in os.listdir(DATASET_PATH):
    if file.startswith("temp_"):
        old_path = os.path.join(DATASET_PATH, file)
        new_name = f"a{final_count}.jpg"  #renaming all like a1,a2,a3... and so on 
        new_path = os.path.join(DATASET_PATH, new_name)

        os.rename(old_path, new_path)
        final_count += 1

print(f"\n✅ Done. Total images: {final_count - 1}")
```

```python
import os
import hashlib

DATASET_PATH = "/Users/prekshyajoshi/instacart_recommender_project/zepto_cnn_dataset_final/good"   # change this

hash_dict = {}   #  to store hashes

duplicates_removed = 0

def get_md5(filepath, chunk_size=6144):  # 6KB chunks
    md5 = hashlib.md5()
    
    try:
        md5 = hashlib.md5()
        
        with open(filepath, "rb") as f:  # open image in binary mode since imgs
            
            while True:
                chunk = f.read(chunk_size)  # read the images in 6kb chunks and repeat this till full img is done

                if not chunk:  
                    break

                md5.update(chunk)  # add this chunk to hash calculation

        return md5.hexdigest()  
        
    except Exception as e:
        print(f"Error reading {filepath}: {e}")
        return None

# Loop through dataset
for root, dirs, files in os.walk(DATASET_PATH):    #here root = "good_dataset" and files = ["img1.jpg", "img2.jpg"]
    for file in files:
        filepath = os.path.join(root, file)  #thisll get the exact path to my image like : good_dataset/img1.jpg

        file_hash = get_md5(filepath)

        if file_hash is None:
            continue

        # Check duplicate
        if file_hash in hash_dict:
            print(f"Duplicate found: {filepath}")
            os.remove(filepath) 
            duplicates_removed += 1
        
        else:
            hash_dict[file_hash] = filepath


print("\n" + "="*40)
print(f"✅ Duplicate removal completed")
print(f"🗑️ Total duplicates removed: {duplicates_removed}")
print("="*40)
```
